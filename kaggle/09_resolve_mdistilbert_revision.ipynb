# Metadata-only mDistilBERT revision resolver
This guarded utility queries Hugging Face repository metadata. It does not import Transformers, download files, or load weights.

```python
RUN_HEAVY = False
EXPECTED_SHA = '45c032ab32cc946ad88a166f7cb282f58c753c2e'

if not RUN_HEAVY:
    print('Disabled. Set RUN_HEAVY=True on Kaggle to query metadata only.')
else:
    import json, re, urllib.request
    url = 'https://huggingface.co/api/models/distilbert/distilbert-base-multilingual-cased'
    with urllib.request.urlopen(url, timeout=30) as response:
        metadata = json.load(response)
    sha = metadata.get('sha', '')
    if not re.fullmatch(r'[0-9a-f]{40}', sha):
        raise RuntimeError(f'Invalid repository SHA returned: {sha!r}')
    print('repository:', metadata.get('modelId'))
    print('immutable_sha:', sha)
    print('matches_committed_stage2_pin:', sha == EXPECTED_SHA)
```
